Очистка явных рейтингов:

Train/test split. Так как времени нет, используем **разбиение по пользователям**:

## 3. Оценка топ-N: метрики и подготовка

Для ranking-метрик нам нужна бинарная “релевантность”.
Определим релевантные как `rating >= threshold` (например, 8).
Также исключаем из рекомендаций всё, что пользователь уже видел в train.

```python
threshold = 8
K = 10

train_items_by_user = train.groupby('user_id')['anime_id'].apply(set)
test_rel_by_user = (
    test[test['rating'] >= threshold]
    .groupby('user_id')['anime_id']
    .apply(set)
)

eval_users = test_rel_by_user.index.intersection(train_items_by_user.index)
len(eval_users)
```

Реализуем Precision@K, Recall@K, MAP@K.

```python
def precision_at_k(recommended, relevant, k):
    rec_k = recommended[:k]
    if len(rec_k) == 0:
        return 0.0
    return float(len([x for x in rec_k if x in relevant]) / k)

def recall_at_k(recommended, relevant, k):
    if len(relevant) == 0:
        return 0.0
    rec_k = recommended[:k]
    return float(len([x for x in rec_k if x in relevant]) / len(relevant))

def apk(recommended, relevant, k):
    rec_k = recommended[:k]
    hit_count = 0
    score = 0.0
    for i, item in enumerate(rec_k, start=1):
        if item in relevant:
            hit_count += 1
            score += hit_count / i
    if len(relevant) == 0:
        return 0.0
    return float(score / min(len(relevant), k))

def mapk(recs_by_user, rel_by_user, k):
    scores = []
    for u in recs_by_user:
        scores.append(apk(recs_by_user[u], rel_by_user[u], k))
    return float(np.mean(scores)) if len(scores) > 0 else 0.0
```

## 4. Модель 1: Baseline (Popular)

Базовая модель: рекомендуем всем пользователям самые популярные аниме по train (по количеству оценок).

```python
popularity = train.groupby('anime_id')['user_id'].size().sort_values(ascending=False)
popular_items = popularity.index.to_list()

def recommend_popular(user_id, k):
    seen = train_items_by_user.get(user_id, set())
    recs = [i for i in popular_items if i not in seen]
    return recs[:k]
```

Оценим модель по метрикам ranking.

```python
recs_pop = {u: recommend_popular(u, K) for u in eval_users}

p_pop = float(np.mean([precision_at_k(recs_pop[u], test_rel_by_user[u], K) for u in eval_users]))
r_pop = float(np.mean([recall_at_k(recs_pop[u], test_rel_by_user[u], K) for u in eval_users]))
map_pop = mapk(recs_pop, test_rel_by_user, K)

p_pop, r_pop, map_pop
```

## 5. Модель 2: Content-Based Filtering

Создаём матрицу признаков товаров:

* TF-IDF по тексту `genre + type`
* добавим численные признаки `episodes`, `rating`, `members` (стандартизируем)

Это и есть “матрица признаков для каждого продукта”.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import StandardScaler
from scipy.sparse import hstack, csr_matrix

items = anime[['anime_id', 'genre', 'type', 'episodes', 'rating', 'members']].copy()
items['genre'] = items['genre'].fillna('')
items['type'] = items['type'].fillna('')

items['text'] = (items['genre'].astype('string') + ' ' + items['type'].astype('string')).fillna('')

tfidf = TfidfVectorizer(min_df=2)
X_text = tfidf.fit_transform(items['text'].to_list())

num = items[['episodes', 'rating', 'members']].copy()
num = num.fillna(num.median(numeric_only=True))

scaler = StandardScaler(with_mean=False)
X_num = scaler.fit_transform(num.astype('float32'))

X_item = hstack([X_text, csr_matrix(X_num)], format='csr')

item_ids_all = items['anime_id'].to_numpy(dtype='int32')
itemid2row = pd.Series(np.arange(len(item_ids_all), dtype='int32'), index=item_ids_all)

X_item.shape
```

Посчитаем **косинусное сходство**:
полная item-item матрица большая, поэтому:

* покажем пример “похожие на выбранное аниме”
* а для рекомендаций будем считать сходство “профиль пользователя ↔ все аниме” (это тоже косинус)

```python
from sklearn.metrics.pairwise import cosine_similarity

sample_anime_id = int(item_ids_all[0])
row = int(itemid2row[sample_anime_id])

sims = cosine_similarity(X_item, X_item[row]).ravel().astype('float32')
top_sim_idx = np.argsort(-sims)[:11]

pd.DataFrame({
    'anime_id': item_ids_all[top_sim_idx],
    'similarity': sims[top_sim_idx]
}).merge(anime[['anime_id', 'name', 'genre', 'type']], on='anime_id', how='left')
```

Реализуем рекомендации Content-Based:

* берём аниме, которые пользователь высоко оценил в train (`>= threshold`)
* строим “профиль пользователя” как средний вектор признаков
* считаем косинусное сходство профиля со всеми аниме
* берём top-K, исключая просмотренное

```python
from sklearn.metrics.pairwise import cosine_similarity

liked_train = (
    train[train['rating'] >= threshold]
    .groupby('user_id')['anime_id']
    .apply(list)
)

def recommend_content(user_id, k):
    liked = liked_train.get(user_id, [])
    if len(liked) == 0:
        return []
    liked = [i for i in liked if i in itemid2row.index]
    if len(liked) == 0:
        return []
    rows = itemid2row.loc[liked].to_numpy(dtype='int32')
    profile = X_item[rows].mean(axis=0)
    profile = np.asarray(profile).ravel()
    scores = cosine_similarity(X_item, profile.reshape(1, -1)).ravel().astype('float32')
    seen = train_items_by_user.get(user_id, set())
    order = np.argpartition(-scores, min(k + 2000, len(scores) - 1))[:min(k + 2000, len(scores))]
    order = order[np.argsort(-scores[order])]
    recs = []
    for idx in order:
        aid = int(item_ids_all[idx])
        if aid not in seen:
            recs.append(aid)
            if len(recs) >= k:
                break
    return recs
```

Оценка Content-Based по Precision@K / Recall@K / MAP@K.

```python
recs_cb = {u: recommend_content(u, K) for u in eval_users}

p_cb = float(np.mean([precision_at_k(recs_cb[u], test_rel_by_user[u], K) for u in eval_users]))
r_cb = float(np.mean([recall_at_k(recs_cb[u], test_rel_by_user[u], K) for u in eval_users]))
map_cb = mapk(recs_cb, test_rel_by_user, K)

p_cb, r_cb, map_cb
```

## 6. Модель 3: Collaborative Filtering (MF + biases)

Реализуем матричную факторизацию с bias
Обучение SGD на явных рейтингах.

Чтобы обучение было выполнимо на CPU, введём фильтрацию “не слишком холодных” пользователей и items.

```python
min_user_ratings_cf = 30
min_item_ratings_cf = 50

user_cnt = train.groupby('user_id')['anime_id'].size()
item_cnt = train.groupby('anime_id')['user_id'].size()

good_users_cf = user_cnt[user_cnt >= min_user_ratings_cf].index
good_items_cf = item_cnt[item_cnt >= min_item_ratings_cf].index

train_cf = train[train['user_id'].isin(good_users_cf) & train['anime_id'].isin(good_items_cf)].copy()
test_cf = test[test['user_id'].isin(good_users_cf) & test['anime_id'].isin(good_items_cf)].copy()

train_cf.shape, test_cf.shape
```

Построим индексы и массивы для обучения.

```python
user_ids_cf = train_cf['user_id'].unique()
item_ids_cf = train_cf['anime_id'].unique()

user2idx_cf = pd.Series(np.arange(len(user_ids_cf), dtype='int32'), index=user_ids_cf)
item2idx_cf = pd.Series(np.arange(len(item_ids_cf), dtype='int32'), index=item_ids_cf)

u = train_cf['user_id'].map(user2idx_cf).to_numpy(dtype='int32')
i = train_cf['anime_id'].map(item2idx_cf).to_numpy(dtype='int32')
y = train_cf['rating'].to_numpy(dtype='float32')

len(user_ids_cf), len(item_ids_cf), len(y)
```

Обучение MF через SGD.
Параметр `max_interactions` ограничивает число взаимодействий (ускорение для ноутбука).

```python
rng = np.random.default_rng(42)

n_factors = 32
lr = 0.02
reg = 0.02
n_epochs = 8
max_interactions = 400000

idx_all = np.arange(len(y))
if len(idx_all) > max_interactions:
    idx_all = rng.choice(idx_all, size=max_interactions, replace=False)
    u_tr = u[idx_all]
    i_tr = i[idx_all]
    y_tr = y[idx_all]
else:
    u_tr, i_tr, y_tr = u, i, y

mu_cf = float(np.mean(y_tr))

P = (0.01 * rng.standard_normal((len(user_ids_cf), n_factors))).astype('float32')
Q = (0.01 * rng.standard_normal((len(item_ids_cf), n_factors))).astype('float32')
bu = np.zeros(len(user_ids_cf), dtype='float32')
bi = np.zeros(len(item_ids_cf), dtype='float32')

for epoch in range(n_epochs):
    order = rng.permutation(len(y_tr))
    u_e = u_tr[order]
    i_e = i_tr[order]
    y_e = y_tr[order]
    se = 0.0
    for uu, ii, rr in zip(u_e, i_e, y_e):
        pred = mu_cf + bu[uu] + bi[ii] + float(np.dot(P[uu], Q[ii]))
        err = rr - pred
        se += err * err
        bu[uu] += lr * (err - reg * bu[uu])
        bi[ii] += lr * (err - reg * bi[ii])
        p_old = P[uu].copy()
        P[uu] += lr * (err * Q[ii] - reg * P[uu])
        Q[ii] += lr * (err * p_old - reg * Q[ii])
    rmse_epoch = float(np.sqrt(se / len(y_e)))
    print(epoch + 1, rmse_epoch)
```

RMSE для MF на test (пересечение по пользователям/аниме, которые есть в train_cf).

```python
test_in = test_cf[test_cf['user_id'].isin(user2idx_cf.index) & test_cf['anime_id'].isin(item2idx_cf.index)].copy()

u_te = test_in['user_id'].map(user2idx_cf).to_numpy(dtype='int32')
i_te = test_in['anime_id'].map(item2idx_cf).to_numpy(dtype='int32')
y_te = test_in['rating'].to_numpy(dtype='float32')

pred_te = mu_cf + bu[u_te] + bi[i_te] + np.sum(P[u_te] * Q[i_te], axis=1)
rmse_cf = float(np.sqrt(np.mean((y_te - pred_te) ** 2)))

rmse_cf, len(test_in)
```

Top-N рекомендации MF:

* считаем скоры для всех items
* исключаем просмотренное
* берём top-K

```python
train_items_by_user_cf = train_cf.groupby('user_id')['anime_id'].apply(set)
all_item_ids_cf = item_ids_cf.astype('int32')

def recommend_cf(user_id, k):
    if user_id not in user2idx_cf.index:
        return []
    uu = int(user2idx_cf[user_id])
    scores = (mu_cf + bu[uu] + bi + (Q @ P[uu])).astype('float32')
    seen = train_items_by_user_cf.get(user_id, set())
    order = np.argpartition(-scores, min(k + 2000, len(scores) - 1))[:min(k + 2000, len(scores))]
    order = order[np.argsort(-scores[order])]
    recs = []
    for idx in order:
        aid = int(all_item_ids_cf[idx])
        if aid not in seen:
            recs.append(aid)
            if len(recs) >= k:
                break
    return recs
```

Оценка MF по ranking-метрикам.
Важно: MF считается на `eval_users_cf` (те, кого покрывает модель после фильтрации).

```python
eval_users_cf = eval_users.intersection(train_items_by_user_cf.index).intersection(test_rel_by_user.index)
len(eval_users_cf)
```

```python
recs_cf = {u: recommend_cf(u, K) for u in eval_users_cf}

p_cf = float(np.mean([precision_at_k(recs_cf[u], test_rel_by_user[u], K) for u in eval_users_cf]))
r_cf = float(np.mean([recall_at_k(recs_cf[u], test_rel_by_user[u], K) for u in eval_users_cf]))
map_cf = mapk(recs_cf, test_rel_by_user.loc[eval_users_cf], K)

p_cf, r_cf, map_cf
```

## 7. Модель 4: Hybrid (Content + Collaborative)

Смешаем скоринги CB и CF
Чтобы шкалы были сопоставимы, применим min-max нормализацию скорингов внутри пользователя.

```python
def minmax_norm(x):
    x = np.asarray(x, dtype='float32')
    mn = float(np.min(x))
    mx = float(np.max(x))
    if mx - mn < 1e-8:
        return np.zeros_like(x)
    return (x - mn) / (mx - mn)
```

Рекомендации Hybrid:

* CF-скор есть всегда для пользователей в CF
* CB-скор строим через профиль пользователя по liked items в train
* объединяем и берём top-K

```python
def recommend_hybrid(user_id, k, alpha):
    if user_id not in user2idx_cf.index:
        return recommend_content(user_id, k)

    uu = int(user2idx_cf[user_id])
    scores_cf = (mu_cf + bu[uu] + bi + (Q @ P[uu])).astype('float32')
    scores_cf = minmax_norm(scores_cf)

    liked = liked_train.get(user_id, [])
    liked = [i for i in liked if i in itemid2row.index]

    if len(liked) == 0:
        scores_cb_full = np.zeros(len(all_item_ids_cf), dtype='float32')
    else:
        rows = itemid2row.loc[liked].to_numpy(dtype='int32')
        profile = X_item[rows].mean(axis=0)
        profile = np.asarray(profile).ravel()
        sim_all = cosine_similarity(X_item, profile.reshape(1, -1)).ravel().astype('float32')

        cb_scores = np.array(
            [sim_all[itemid2row[aid]] if aid in itemid2row.index else 0.0 for aid in all_item_ids_cf],
            dtype='float32'
        )
        scores_cb_full = minmax_norm(cb_scores)

    scores = alpha * scores_cf + (1.0 - alpha) * scores_cb_full

    seen = train_items_by_user_cf.get(user_id, set())
    order = np.argpartition(-scores, min(k + 2000, len(scores) - 1))[:min(k + 2000, len(scores))]
    order = order[np.argsort(-scores[order])]

    recs = []
    for idx in order:
        aid = int(all_item_ids_cf[idx])
        if aid not in seen:
            recs.append(aid)
            if len(recs) >= k:
                break
    return recs
```

Подбор `alpha` по MAP@K на подмножестве пользователей (чтобы быстрее).

```python
eval_users_h = list(eval_users_cf)
rng = np.random.default_rng(42)

if len(eval_users_h) > 3000:
    eval_users_h = rng.choice(eval_users_h, size=3000, replace=False).tolist()

alphas = np.linspace(0, 1, 11)
rows = []
for a in alphas:
    recs_h = {u: recommend_hybrid(u, K, float(a)) for u in eval_users_h}
    map_h = mapk(recs_h, test_rel_by_user.loc[eval_users_h], K)
    rows.append({'alpha': float(a), 'map@k': float(map_h)})

df_alpha = pd.DataFrame(rows)
best_alpha = float(df_alpha.sort_values('map@k', ascending=False).iloc[0]['alpha'])

df_alpha, best_alpha
```

```python
fig = px.line(df_alpha, x='alpha', y='map@k', markers=True, title='Подбор alpha по MAP@K')
fig.show()
```

Оценка Hybrid модели по ranking-метрикам.

```python
recs_hybrid = {u: recommend_hybrid(u, K, best_alpha) for u in eval_users_cf}

p_h = float(np.mean([precision_at_k(recs_hybrid[u], test_rel_by_user[u], K) for u in eval_users_cf]))
r_h = float(np.mean([recall_at_k(recs_hybrid[u], test_rel_by_user[u], K) for u in eval_users_cf]))
map_h = mapk(recs_hybrid, test_rel_by_user.loc[eval_users_cf], K)

p_h, r_h, map_h
```

## 8. Сравнение моделей (таблица/график) + выводы

Сводим результаты в таблицу и строим график сравнения по MAP@K.
Также добавим RMSE для MF (и для Hybrid можно оставить тот же RMSE как для CF части).

```python
results = []
results.append({'model': 'Popular baseline', 'precision@k': p_pop, 'recall@k': r_pop, 'map@k': map_pop, 'rmse': np.nan, 'users': int(len(eval_users))})
results.append({'model': 'Content-based', 'precision@k': p_cb, 'recall@k': r_cb, 'map@k': map_cb, 'rmse': np.nan, 'users': int(len(eval_users))})
results.append({'model': 'Collaborative MF', 'precision@k': p_cf, 'recall@k': r_cf, 'map@k': map_cf, 'rmse': float(rmse_cf), 'users': int(len(eval_users_cf))})
results.append({'model': f'Hybrid (alpha={best_alpha:.2f})', 'precision@k': p_h, 'recall@k': r_h, 'map@k': map_h, 'rmse': float(rmse_cf), 'users': int(len(eval_users_cf))})

df_results = pd.DataFrame(results)
df_results
```

```python
fig = px.bar(df_results, x='model', y='map@k', title='Сравнение моделей по MAP@K')
fig.update_layout(xaxis_tickangle=-30)
fig.show()
```

Определим лучшую модель по MAP@K и подготовим наглядный пример для одного пользователя:

**Результаты сравнения моделей.**
По полученным метрикам лучшей оказалась модель **Popular baseline** (MAP@10 ≈ 0.052), далее **Content-based** (MAP@10 ≈ 0.020). **Collaborative MF** и **Hybrid** показали более низкие значения MAP@10 на текущей настройке эксперимента.

**Почему так могло получиться:**

1. **Сильная разреженность данных.** Матрица user–item крайне разреженная, что усложняет обучение коллаборативных моделей и повышает роль “глобальной популярности”.
2. **Определение релевантности через высокий порог** (`rating >= 8`). При таком пороге релевантных объектов у пользователя обычно мало, поэтому модели, которые “угадывают” мейнстрим-популярное, могут выигрывать в MAP@K.
3. **Ограничение покрытия пользователей в CF.** Коллаборативная модель обучалась на подмножестве пользователей/аниме (фильтрация по минимальному числу оценок), поэтому метрики считались на меньшем числе пользователей. Это снижает сравнимость с baseline/content на полном покрытии и дополнительно ухудшает итоговые значения.
4. **Ограниченность контентных признаков.** Content-based модель использовала в основном жанры и тип (и несколько численных полей). Если контентных признаков мало, модель не может “поймать” тонкие предпочтения, но она всё равно остаётся полезной для cold-start.

**Пример рекомендаций лучшей модели.**
Для выбранного пользователя была показана история его оценок (train) и список топ-10 рекомендаций. В случае Popular baseline рекомендации объясняются тем, что эти аниме наиболее популярны среди большинства пользователей (высокое число взаимодействий в train). В случае content-based рекомендации формируются за счёт схожести жанров/типа с теми аниме, которые пользователь высоко оценил.

**Ограничения работы.**

**Рекомендации по улучшению:**

* Выделить отдельный **validation**-набор из train и подбирать гиперпараметры без использования test.
* Для collaborative части попробовать более устойчивые подходы: ALS (implicit), BPR, LightFM, а также учесть `rating = -1` как **неявные** взаимодействия.
